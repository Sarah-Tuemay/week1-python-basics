# Week 1 Lab — Student Grade Data Processing

This notebook loads a raw CSV of student grades, cleans it with functions,
processes it with NumPy, and prints a summary report.

Tools: Python, Jupyter, NumPy, venv

```python
import numpy as np
import data_loader
import analyzer
```

## 1. Load and clean the raw data

`load_clean_data` reads `data/students.csv`, marks unconvertible grades as
missing, fills each missing subject with that subject's average, removes any
unfixable rows, and drops duplicate students.

```python
records = data_loader.load_clean_data("data/students.csv")

print(f"Clean records: {len(records)}")
for record in records:
    print(record)
```

## 2. Build the score matrix

Convert the list of records into a 2D NumPy array: one row per student,
one column per subject (math, physics, english).

```python
scores = analyzer.get_scores(records)
print(scores)
print("Shape:", scores.shape)
```

## 3. Process with NumPy

Compute subject averages, the top student, and the pass rate.

```python
print("Subject averages:", analyzer.subject_averages(scores))
print("Top student:", analyzer.top_student(records, scores))
print("Pass rate (threshold 85):", analyzer.pass_rate(scores, 85), "%")
```

## Results summary

- The dataset had 10 rows with 3 dirty-data problems: one missing value
  (Liya, physics), one non-numeric value (Kebede, `N/A`), and one duplicate
  (Abebe).
- After cleaning: 9 unique, complete records.
- All missing physics scores were filled with the mean of known physics scores.
